## Identify trends, distributions, and relationships between variables.

```python
# Calculate total sales

total_sales = data['order_total'].sum()

print("order_total:", total_sales)
```

```python
# Calculate average sales per transaction

average_sales = data['order_total'].mean()

print("Average Sales per Transaction:", round(average_sales, 2))
```

```python
# Calculate total quantity sold

total_quantity = data['quantity'].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
# Calculate total sales for each category

category_sales = (
    data.groupby('category')['order_total']
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)
```

```python
# Identify the category with the lowest total sales

lowest_category = category_sales.idxmin()
lowest_category_sales = category_sales.min()

print("Lowest Sales Category:", lowest_category)
print("Sales:", lowest_category_sales)
```

```python
# Calculate total sales for each product

product_sales = (
    data.groupby('item')['order_total']
    .sum()
    .sort_values(ascending=False)
)

product_sales.head(10)
```

```python
# Display the top 10 products based on total sales

top_10_products = product_sales.head(10)

print("Top 10 Products:")
print(top_10_products)
```

```python
# Count transactions by payment method

payment_count = data['payment_method'].value_counts()

print(payment_count)
```

```python
# Calculate total sales for each payment method

payment_sales = (
    data.groupby('payment_method')['order_total']
    .sum()
    .sort_values(ascending=False)
)

print(payment_sales)
```

```python
# Identify the most frequently used payment method

most_used_payment = data['payment_method'].value_counts().idxmax()

print("Most Popular Payment Method:", most_used_payment)
```

## Detect outliers and unusual patterns within the data.

```python
# Calculate correlation between numerical variables

correlation = data[numeric_columns].corr()

correlation
```

```python
# Calculate the correlation between Quantity and Total Sales

quantity_sales_correlation = data['quantity'].corr(
    data['order_total']
)

print(
    "Correlation between Quantity and Total Sales:",
    round(quantity_sales_correlation, 2)
)
```

```python
# Create a heatmap to visualize relationships
# between numerical variables

plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Calculate Q1 and Q3 for Total Sales

Q1 = data['order_total'].quantile(0.25)
Q3 = data['order_total'].quantile(0.75)

# Calculate Interquartile Range
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
# Calculate lower and upper limits

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Identify outliers
outliers = data[
    (data['order_total'] < lower_limit) |
    (data['order_total'] > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Outliers:", len(outliers))
```

```python
# Display the first few outlier records

outliers.head(10)
```

```python
# Display transactions with the highest sales

data.sort_values(
    by='order_total',
    ascending=False
).head(10)
```

##  Use summary statistics to answer key business questions.

```python
# Calculate important business metrics

total_sales = data["order_total"].sum()

average_order_value = data["order_total"].mean()

highest_order = data["order_total"].max()

lowest_order = data["order_total"].min()

total_orders = data["order_id"].nunique()

total_customers = data["customer_id"].nunique()

total_quantity = data["quantity"].sum()
```

```python
# Calculate values for business questions

# 1. Total Sales
total_sales = data["order_total"].sum()

# 2. Average Sales
average_sales = data["order_total"].mean()

# 3. Category-wise Sales
category_sales = data.groupby("category")["order_total"].sum()

highest_category = category_sales.idxmax()
lowest_category = category_sales.idxmin()

# 4. Product-wise Sales
product_sales = data.groupby("item")["order_total"].sum()

# 5. Most Used Payment Method
most_used_payment = data["payment_method"].value_counts().idxmax()

# 6. Quantity-Sales Correlation
quantity_sales_correlation = data["quantity"].corr(data["order_total"])

# 7. Sales Outliers using IQR
Q1 = data["order_total"].quantile(0.25)
Q3 = data["order_total"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[
    (data["order_total"] < lower_bound) |
    (data["order_total"] > upper_bound)
]
```

```python
# Answer important business questions

print("===== BUSINESS INSIGHTS =====")

print("1. Total Sales:", round(total_sales, 2))

print(
    "2. Average Sales:",
    round(average_sales, 2)
)

print(
    "3. Highest Sales Category:",
    highest_category
)

print(
    "4. Lowest Sales Category:",
    lowest_category
)

print(
    "5. Top Product:",
    product_sales.idxmax()
)

print(
    "6. Most Used Payment Method:",
    most_used_payment
)

print(
    "7. Quantity-Sales Correlation:",
    round(quantity_sales_correlation, 2)
)

print(
    "8. Number of Sales Outliers:",
    len(outliers)
)
```
