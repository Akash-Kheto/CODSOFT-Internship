## Clean the dataset by handling null values, removing duplicates, and correcting data types.

```python
# Create a copy for cleaning

cleaned_data = data.copy()

print("Copy created successfully!")
```

```python
# Remove leading and trailing spaces from text columns
text_columns = cleaned_data.select_dtypes(include="object").columns
for column in text_columns:
    cleaned_data[column] = cleaned_data[column].str.strip()

print("Extra spaces removed successfully!")
```

```python
# Standardize column names

cleaned_data.columns = (
    cleaned_data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Updated Column Names:")
print(cleaned_data.columns)
```

```python
# Check missing values before handling them

print(cleaned_data.isnull().sum())
```

```python
# Fill numerical missing values with median
numeric_columns = cleaned_data.select_dtypes(include=np.number).columns
for column in numeric_columns:
    cleaned_data[column] = cleaned_data[column].fillna(
        cleaned_data[column].median()
    )
```

```python
# Fill missing categorical/text values

categorical_columns = cleaned_data.select_dtypes(
    include="object"
).columns

for column in categorical_columns:
    cleaned_data[column] = cleaned_data[column].fillna("Unknown")

print("Missing categorical values handled!")
```

```python
# Remove duplicate records

before_duplicates = len(cleaned_data)

cleaned_data = cleaned_data.drop_duplicates()

after_duplicates = len(cleaned_data)

print("Rows before removing duplicates:", before_duplicates)
print("Rows after removing duplicates:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)
```

## Prepare the data for further analysis using Pandas.

```python
# Convert Price to numeric

cleaned_data["price"] = pd.to_numeric(
    cleaned_data["price"],
    errors="coerce"
)
```

```python
# Convert Quantity to numeric

cleaned_data["quantity"] = pd.to_numeric(
    cleaned_data["quantity"],
    errors="coerce"
)
```

```python
# Convert Quantity to integer

cleaned_data["quantity"] = cleaned_data["quantity"].round().astype("int64")
```

```python
# Convert Order Total to numeric

cleaned_data["order_total"] = pd.to_numeric(
    cleaned_data["order_total"],
    errors="coerce"
)

# Fill missing values

cleaned_data["order_total"] = cleaned_data["order_total"].fillna(
    cleaned_data["order_total"].median()
)
```

```python
# Standardize Category values

cleaned_data["category"] = (
    cleaned_data["category"]
    .str.strip()
    .str.title()
)

print(cleaned_data["category"].value_counts())
```

```python
# Standardize Payment Method values

cleaned_data["payment_method"] = (
    cleaned_data["payment_method"]
    .str.strip()
    .str.title()
)

print(cleaned_data["payment_method"].value_counts())
```

```python
# Check missing values after cleaning

print("Missing Values After Cleaning:")
print(cleaned_data.isnull().sum())
```

```python
# Check duplicates after cleaning

print(
    "Duplicate records after cleaning:",
    cleaned_data.duplicated().sum()
)
```
